## Data Exploration

The CIFAR-10 dataset contains 10 classes of images.

Each image has a size of 32x32 pixels with 3 color channels (RGB).

The dataset is balanced, meaning that each class contains approximately the same number of images.

This makes it suitable for training classification models without class imbalance issues.

```python
import torch
import torchvision
import torchvision.transforms as transforms
import matplotlib.pyplot as plt
```

```python
transform = transforms.ToTensor()

trainset = torchvision.datasets.CIFAR10(
    root='./data',
    train=True,
    download=True,
    transform=transform
)
```

```python
classes = trainset.classes

for i in range(5):
    img, label = trainset[i]
    plt.imshow(img.permute(1, 2, 0))
    plt.title(classes[label])
    plt.axis("off")
    plt.show()
```

### Class Distribution

The dataset is balanced across all classes, which helps the model learn without bias toward a specific category.

```python
from collections import Counter

labels = [label for _, label in trainset]
counter = Counter(labels)

plt.bar(counter.keys(), counter.values())
plt.title("Class Distribution")
plt.show()
```

### Conclusion

The CIFAR-10 dataset is balanced and suitable for classification tasks.

However, the small image size (32x32) may limit model performance and make some classes harder to distinguish.
